그래프 탐색 실습(DFS, BFS, 친구 추천)

```python
#예시 그래프 : 친구 관계
graph={
    'Alice':["Bob", 'Charlie'],
    'Bob':['David', 'Eve'],
    'Charlie':['Frank'],
    'David':[],
    'Eve':['Frank'],
    'Frank':[]
    }
```

DFS 구현

```python
def dfs(graph, v, visited):
  visited.add(v),
  print(v,end='')
  for neighbor in graph[v]:
    if neighbor not in visited:
      if neighbor not in visited:
        dfs(graph, neighbor, visited)

print("DFS 탐색 순서:")
dfs(graph, 'Alice', set())
```

BFS 탐색 구현

```python
#큐를 효율적으로 구현할 수 있는 도구
from collections import deque
def bfs(graph, start):
  visited=set()
  queue=deque([start])
  visited.add(start)

  while queue:
    v=queue.popleft()
    print(v, end='')
    for neighbor in graph[v]:
      if neighbor not in visited:
        visited.add(neighbor)
        queue.append(neighbor)

print('\nBFS 탐색 순서 :')
bfs(graph, 'Alice')
```
